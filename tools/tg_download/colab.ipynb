# Kanal materiallarini Drive'ga yuklash
Telegram kanalidagi fayllar Google serverida yuklanadi va Drive'ga yoziladi (sizning internetingiz sarflanmaydi).
Avval chap paneldagi **Secrets** (kalit belgisi) bo'limiga `TG_API_ID` va `TG_API_HASH` ni qo'shing va "Notebook access" ni yoqing.
Kataklarni tartib bilan ishga tushiring.

```python
# 1. Drive, kalitlar va Telethon (cryptg: shifrlashni ~10x tezlashtiradi, busiz yuklash juda sekin)
%pip install -q telethon cryptg
import os, sys
from google.colab import drive, userdata, files

drive.mount('/content/drive')
os.environ['TG_API_ID'] = userdata.get('TG_API_ID')
os.environ['TG_API_HASH'] = userdata.get('TG_API_HASH')
os.environ['TG_OUT'] = '/content/drive/MyDrive/ilmbot/channel_dump'   # natija Drive'da
os.environ['TG_SESSION'] = '/content/telegram'                         # sessiya Drive'ga TUSHMAYDI
os.environ['TG_TMP'] = '/content/tmp'                                  # yarim fayllar tez lokal diskda
```

```python
# 2. Skriptni yuklang: download_channel.py (loyihadagi tools/tg_download/ papkasidan)
files.upload()
sys.path.insert(0, '/content')
import download_channel as dc
```

```python
# 3. Telegramga kirish: telefon raqami (+998...) va Telegramdan kelgan kodni kiriting
await dc.login()
```

```python
# 4. Sinov: dastlabki 5 ta post
await dc.download('@testislomyolida', 5)
```

```python
# 5. Hammasini yuklash (~25,5 GB). Uzilsa shu katakni qayta ishga tushiring: tayyor fayllar o'tkazib yuboriladi
# workers: bir vaqtda nechta fayl. FloodWait ko'payса 4 ga tushiring
await dc.download('@testislomyolida', None, workers=8)
```

```python
# 6. Tugagach: sessiyani o'chiring, so'ng Telegram > Sozlamalar > Qurilmalar'dan Colab seansini tugating
import glob
for f in glob.glob('/content/telegram*'):
    os.remove(f)
print('Sessiya o'chirildi')
```
